```python
import pandas as pd
import numpy as np
import requests
from io import StringIO
```

```python
urls = {
    "adult_literacy": "https://ourworldindata.org/grapher/literacy-rate-adults.csv?v=1&csvType=full&useColumnShortNames=true",

    "youth_literacy": "https://ourworldindata.org/grapher/literacy-rate-of-young-men-and-women.csv?v=1&csvType=full&useColumnShortNames=true",

    "illiteracy": "https://ourworldindata.org/grapher/literate-and-illiterate-world-population.csv?v=1&csvType=full&useColumnShortNames=true",

    "gdp": "https://ourworldindata.org/grapher/gdp-per-capita-worldbank.csv?v=1&csvType=full&useColumnShortNames=true",

    "schooling": "https://ourworldindata.org/grapher/literacy-rates-vs-average-years-of-schooling.csv?v=1&csvType=full&useColumnShortNames=true"
}
```

```python
datasets = {}

for name, url in urls.items():
    print(f"Downloading {name}...")

    response = requests.get(url)

    if response.status_code == 200:
        datasets[name] = pd.read_csv(StringIO(response.text))
        print(f"{name} downloaded successfully.")
    else:
        print(f"Failed to download {name}")
```

```python
for name, df in datasets.items():
    print("=" * 50)
    print(name)
    print(df.shape)
    print(df.head())
```

```python
adult = datasets["adult_literacy"]
youth = datasets["youth_literacy"]
illiteracy = datasets["illiteracy"]
gdp = datasets["gdp"]
schooling = datasets["schooling"]
```

```python
def clean_columns(df):
    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
        .str.replace(" ", "_")
        .str.replace("-", "_")
    )
    return df

adult = clean_columns(adult)
youth = clean_columns(youth)
illiteracy = clean_columns(illiteracy)
gdp = clean_columns(gdp)
schooling = clean_columns(schooling)
```

```python
adult = adult.drop_duplicates()
youth = youth.drop_duplicates()
illiteracy = illiteracy.drop_duplicates()
gdp = gdp.drop_duplicates()
schooling = schooling.drop_duplicates()
```

```python
adult = adult.fillna(method="ffill")
youth = youth.fillna(method="ffill")
illiteracy = illiteracy.fillna(method="ffill")
gdp = gdp.fillna(method="ffill")
schooling = schooling.fillna(method="ffill")
```

```python
adult = adult[(adult["year"] >= 1990) & (adult["year"] <= 2023)]
youth = youth[(youth["year"] >= 1990) & (youth["year"] <= 2023)]
illiteracy = illiteracy[(illiteracy["year"] >= 1990) & (illiteracy["year"] <= 2023)]
gdp = gdp[(gdp["year"] >= 1990) & (gdp["year"] <= 2023)]
schooling = schooling[(schooling["year"] >= 1990) & (schooling["year"] <= 2023)]
```

Feature Engineering

```python
df_literacy["youth_literacy_average"] = (
    df_literacy["youth_literacy_male"] +
    df_literacy["youth_literacy_female"]
) / 2
```

```python
df_literacy["literacy_gender_gap"] = (
    df_literacy["youth_literacy_male"] -
    df_literacy["youth_literacy_female"]
).abs()
```

```python
df_gdp_schooling["gdp_per_schooling_year"] = (
    df_gdp_schooling["gdp_per_capita"] /
    df_gdp_schooling["avg_years_schooling"]
)
```

```python
df_gdp_schooling["gdp_per_schooling_year"] = (
    df_gdp_schooling["gdp_per_capita"] /
    df_gdp_schooling["avg_years_schooling"].replace(0, np.nan)
)
```

```python
df_gdp_schooling["education_index"] = (
    (df_gdp_schooling["literacy_rate"] / 100) *
    (df_gdp_schooling["avg_years_schooling"] / 15)
)
```

```python
df_literacy = df_literacy.sort_values(["country","year"])
```

```python
df_literacy["literacy_growth_rate"] = (
    df_literacy.groupby("country")["adult_literacy_rate"]
    .pct_change() * 100
)
```

```python
df_illiteracy["illiteracy_percentage"] = df_illiteracy["illiteracy_rate"]
```

```python
df_literacy["literacy_improvement_score"] = (
    df_literacy["adult_literacy_rate"] -
    df_literacy["youth_literacy_average"]
)
```

```python
df_gdp_schooling["schooling_category"] = pd.cut(
    df_gdp_schooling["avg_years_schooling"],
    bins=[0,5,8,12,20],
    labels=[
        "Low",
        "Medium",
        "High",
        "Very High"
    ]
)
```

```python
df_gdp_schooling["gdp_category"] = pd.qcut(
    df_gdp_schooling["gdp_per_capita"],
    q=4,
    labels=[
        "Low Income",
        "Lower Middle",
        "Upper Middle",
        "High Income"
    ]
)
```

```python
df_literacy["literacy_level"] = pd.cut(
    df_literacy["adult_literacy_rate"],
    bins=[0,60,80,90,100],
    labels=[
        "Low",
        "Moderate",
        "High",
        "Excellent"
    ]
)
```

```python
df_literacy["top_performer"] = np.where(
    df_literacy["adult_literacy_rate"] >= 95,
    "Yes",
    "No"
)
```
