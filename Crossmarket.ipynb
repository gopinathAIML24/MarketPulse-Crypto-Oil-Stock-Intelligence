```python
import requests
import pandas as pd

coin_ids = ["bitcoin", "ethereum", "tether"]

all_data = []

for coin_id in coin_ids:
    url = f"https://api.coingecko.com/api/v3/coins/{coin_id}/market_chart?vs_currency=inr&days=365"

    response = requests.get(url)
    print(response.status_code)

    if response.status_code == 200:
        data = response.json()

        # Create DataFrame from prices list
        df = pd.DataFrame(data["prices"], columns=["timestamp", "price_inr"])

        # Convert milliseconds to date
        df["date"] = pd.to_datetime(df["timestamp"], unit="ms").dt.date

        # Add coin id
        df["coin_id"] = coin_id

        # Keep only required columns in required order
        df = df[["coin_id", "date", "price_inr"]]

        # Append to list
        all_data.append(df)

# Combine all DataFrames
final_df = pd.concat(all_data, ignore_index=True)

# Format INR values
final_df["price_inr"] = final_df["price_inr"].map(lambda x: f"₹{x:,.2f}")

# Display without index
print(final_df.to_string(index=False))
```

```python
import sqlite3

# Connect to database
conn = sqlite3.connect("crypto.db")

# Insert DataFrame into table
final_df.to_sql(
    "crypto_price_history",   # table name
    conn,
    if_exists="replace",      # use "append" if table already exists and you want to add rows
    index=False
)

# Save changes
conn.commit()

# Close connection
conn.close()

print("Data inserted successfully into crypto_price_history table.")
```

```python
import pandas as pd

# Load data
oil_df = pd.read_csv(
    "https://raw.githubusercontent.com/datasets/oil-prices/main/data/wti-daily.csv"
)

# Convert Date column to datetime
oil_df["Date"] = pd.to_datetime(oil_df["Date"])

# Filter from Jan 2020 to Jun 2026
oil_df = oil_df[
    (oil_df["Date"] >= "2020-01-01") &
    (oil_df["Date"] <= "2026-06-30")
]
#display(oil_df)
print(oil_df.to_string(index=False))
```
